### Idea clave

- Una **lista de diccionarios** puede representar una base de datos pequeña.

## 6. Filtrar registros con condiciones

Ahora combinamos tres elementos ya conocidos:

1. la lista contiene toda la base;
2. el ciclo revisa un diccionario a la vez;
3. la condición decide qué registros mostrar.

Antes de ejecutar, estima cuántas personas aparecerán.

````python
for estudiante in estudiantes:
    if estudiante["asistencia"] >= 80:
        print(estudiante["nombre"], estudiante["asistencia"])
````

### Crear una nueva lista a partir de un filtro

En lugar de solamente imprimir los resultados, podemos guardarlos en otra estructura.

````python
estudiantes_quito = []

for estudiante in estudiantes:
    if estudiante["ciudad"] == "Quito":
        estudiantes_quito.append(estudiante)

print("Cantidad:", len(estudiantes_quito))
print(estudiantes_quito)
````

### Dos condiciones

Predice quiénes cumplen simultáneamente las dos condiciones. Después ejecuta y verifica.

> Los umbrales no son neutrales ni aparecen por sí solos: alguien los define para un propósito. Siempre debemos poder explicar por qué usamos 6 horas o 80 % y revisar a quiénes incluye o excluye esa decisión.

````python
seleccionados = []

for estudiante in estudiantes:
    if estudiante["horas_estudio"] >= 6 and estudiante["asistencia"] >= 80:
        seleccionados.append(estudiante)

for estudiante in seleccionados:
    print(estudiante["nombre"])

print("Total:", len(seleccionados))
````
