## 4. ¿Cuántos datos faltan?

Un dato faltante no es necesariamente cero. La función `isna()` identifica ausencias; `sum()` cuenta cuántas hay en cada columna.

```python
faltantes = df.isna().sum().sort_values(ascending=False)
faltantes[faltantes > 0]
```

```python
print("Número Alumnos faltantes:", df["Número Alumnos"].isna().sum())
print("Número Alumnos igual a cero:", (df["Número Alumnos"] == 0).sum())
print("Registros con valor no faltante:", df["Número Alumnos"].count())
```

**Para discutir:** ¿qué diferencia hay entre una celda vacía y una institución con cero estudiantes registrados? No cambies automáticamente los vacíos por cero: eso alteraría algunas medidas. Si una cifra parece extraña, revisa la documentación y el contexto antes de corregirla.

## 5. Medidas de centro: promedio y mediana

- **Promedio (media):** suma de los valores dividida para la cantidad de valores no faltantes.
- **Mediana:** valor central cuando ordenamos los datos. La mitad de los valores está a cada lado (o tan cerca como permite el tamaño de la muestra).
- **Moda:** valor que se repite con más frecuencia; puede haber más de una.

`pandas` omite los valores faltantes en `mean()` y `median()`. **Predice:** si hay unas pocas instituciones muy grandes, ¿qué medida crees que será mayor?

```python
alumnos = df["Número Alumnos"]
print("Promedio:", round(alumnos.mean(), 2))
print("Mediana:", alumnos.median())
print("Moda(s):", alumnos.mode().tolist()[:10])
```

## 6. Medidas de dispersión: no todos los registros son iguales

El mínimo y el máximo indican extremos. Los cuartiles sitúan el 25 %, 50 % y 75 % de los valores ordenados. El **rango intercuartílico (RIC)** es `Q3 - Q1`: describe la amplitud de la mitad central de los datos. El máximo puede verse muy afectado por un valor extremo; el RIC ayuda a mirar el centro de la distribución.

```python
q1 = alumnos.quantile(0.25)
q3 = alumnos.quantile(0.75)

print("Mínimo:", alumnos.min())
print("Q1 (25 %):", q1)
print("Mediana (50 %):", alumnos.median())
print("Q3 (75 %):", q3)
print("Máximo:", alumnos.max())
print("Rango intercuartílico:", q3 - q1)
```

```python
alumnos.describe().round(2)
```

**Lee `describe()`:** `count` es la cantidad de valores no faltantes, `mean` es el promedio, `std` es la desviación estándar, y `25%`, `50%`, `75%` son cuartiles. No necesitas calcular la desviación estándar a mano hoy: concéntrate en explicar qué representan los valores y en comparar medidas.

## 8. Estadística descriptiva para categorías

Un nombre de provincia o una categoría de sostenimiento no tiene promedio. Para estas variables, hacemos **conteos** y **porcentajes**. Un porcentaje depende de su denominador: aquí será el total de registros con dato no faltante en `Sostenimiento`.

```python
conteos = df["Sostenimiento"].value_counts()
porcentajes = df["Sostenimiento"].value_counts(normalize=True).mul(100).round(1)

resumen_sostenimiento = pd.DataFrame({
    "Instituciones": conteos,
    "Porcentaje entre registros no faltantes": porcentajes
})
resumen_sostenimiento
```

## 9. Comparar grupos: ¿qué cambia entre provincias?

`groupby()` reúne las instituciones por provincia y `agg()` calcula varias medidas a la vez. Como práctica guiada, miraremos las provincias con **más instituciones registradas** y sus valores de estudiantes. Una provincia con pocas instituciones puede tener un promedio inestable, así que mostramos el tamaño del grupo.

```python
resumen_provincia = (
    df.groupby("Provincia")
      .agg(
          instituciones=("Provincia", "size"),
          con_dato_alumnos=("Número Alumnos", "count"),
          promedio_alumnos=("Número Alumnos", "mean"),
          mediana_alumnos=("Número Alumnos", "median")
      )
      .sort_values("instituciones", ascending=False)
)

resumen_provincia.head(10).round(1)
```
