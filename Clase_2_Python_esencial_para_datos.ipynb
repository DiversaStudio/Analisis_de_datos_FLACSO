# 2. Comparaciones: preguntas que Python responde

Una comparación pregunta por la relación entre dos valores. El resultado es un valor lógico: `True` o `False`.

| Operador | Significado |
|---|---|
| `>` | mayor que |
| `<` | menor que |
| `>=` | mayor o igual que |
| `<=` | menor o igual que |
| `==` | igual a |
| `!=` | diferente de |

> Recuerda: `=` asigna un valor; `==` compara dos valores.

# 3. Condicionales: tomar decisiones

Un condicional permite ejecutar una instrucción solamente cuando se cumple una regla.

- `if`: si se cumple una condición;
- `elif`: si se cumple otra condición;
- `else`: en cualquier otro caso.

La **sangría** indica qué instrucciones pertenecen a cada bloque.


```python
participantes = 24
meta = 30

if participantes >= meta:
    print("La meta fue alcanzada")
else:
    print("La meta todavía no fue alcanzada")
```

# 5. Ciclos `for`: repetir una operación

Un ciclo `for` permite recorrer una colección y aplicar una instrucción a cada elemento.


```python
calificaciones = [8.5, 6.2, 9.1, 7.4, 5.8]

for calificacion in calificaciones:
    print("Ahora estoy revisando:", calificacion)
```

En cada repetición, la variable `calificacion` contiene un elemento diferente. La lista no cambia; cambia el valor temporal que estamos revisando.


### Combinar un ciclo con un condicional


```python
for calificacion in calificaciones:
    if calificacion >= 7:
        print(calificacion, "- Aprobado")
    else:
        print(calificacion, "- No aprobado")
```

## Recorrer dos listas relacionadas

`zip()` empareja temporalmente los elementos que ocupan la misma posición en dos listas.


```python
nombres = ["Ana", "Luis", "María", "Carlos", "Elena"]
horas_estudio = [5, 3, 6, 2, 4]

for nombre, horas in zip(nombres, horas_estudio):
    print(nombre, horas)
```

## Contar resultados

Un contador comienza en cero y aumenta cada vez que se cumple una condición.


```python
personas_con_meta = 0

for horas in horas_estudio:
    if horas >= 4:
        personas_con_meta = personas_con_meta + 1

print("Personas que alcanzaron la meta:", personas_con_meta)
```

### Del número al porcentaje


```python
total_personas = len(horas_estudio)
porcentaje = personas_con_meta / total_personas * 100

print(f"{personas_con_meta} de {total_personas} personas alcanzaron la meta.")
print(f"Esto representa el {porcentaje}% del grupo.")
```

# 6. Funciones: reutilizar un procedimiento

Una función recibe información, ejecuta una tarea y puede devolver un resultado.


```python
def calcular_promedio(datos):
    promedio = sum(datos) / len(datos)
    return promedio
```

Definir una función no produce todavía un resultado. Para utilizarla, escribimos su nombre y entregamos los datos entre paréntesis.


```python
horas_estudio = [5, 3, 6, 2, 4]
resultado = calcular_promedio(horas_estudio)

print("Promedio de horas:", resultado)
```

### Reutilizar la misma función


```python
edades = [32, 27, 35, 41, 29]
calificaciones = [8.5, 6.2, 9.1, 7.4, 5.8]

print("Promedio de edades:", calcular_promedio(edades))
print("Promedio de calificaciones:", calcular_promedio(calificaciones))
```

```python
# Desafío final - Resultados generales
horas = []
personas_con_meta = 0

for persona in participantes:
    horas.append(persona["horas"])
    if persona["horas"] >= 4:
        personas_con_meta += 1

promedio_horas = calcular_promedio(horas)
minimo_horas = min(horas)
maximo_horas = max(horas)
porcentaje_meta = personas_con_meta / len(participantes) * 100

print("Promedio:", promedio_horas)
print("Mínimo:", minimo_horas)
print("Máximo:", maximo_horas)
print("Personas que alcanzaron la meta:", personas_con_meta)
print("Porcentaje:", porcentaje_meta)
```
