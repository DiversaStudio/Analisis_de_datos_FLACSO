````python
nombre = "Ana"
edad = 32
estatura = 1.65
participa = True

print(type(nombre))
print(type(edad))
print(type(estatura))
print(type(participa))
````

# 5. Listas: varios valores en una variable

Una lista permite guardar una colección ordenada de valores. Los elementos se escriben entre corchetes y se separan con comas.

````python
edades = [25, 31, 42, 28]

print(edades)
print("Cantidad de personas:", len(edades))
print("Suma de las edades:", sum(edades))
print("Edad mínima:", min(edades))
print("Edad máxima:", max(edades))
````

Python comienza a contar las posiciones desde cero. Por eso, el primer elemento está en la posición `0`.

````python
print(edades[0])
print(edades[1])
````

Ahora convertiremos una colección de valores en una tabla. Usaremos **Pandas**, una herramienta de Python especializada en el análisis de datos. No necesitas memorizar todavía cada parte de este código.

````python
import pandas as pd

datos = {
    "nombre": ["Ana", "Luis", "María", "Carlos", "Elena"],
    "edad": [32, 27, 35, 41, 29],
    "ciudad": ["Quito", "Cuenca", "Quito", "Guayaquil", "Cuenca"],
    "horas_estudio": [5, 3, 6, 2, 4],
}

df = pd.DataFrame(datos)
df
````
